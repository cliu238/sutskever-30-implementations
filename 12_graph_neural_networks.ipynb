图神经网络(graph neural networks)的统一框架，是现代 GNN 的基石！

## 📖 论文导读

**🎯 这篇文章想解决什么问题（目的）：** 化学家想预测一个分子的性质（比如毒性、稳定性），传统方法要么靠昂贵的量子力学计算（一个分子算几个小时），要么靠人工设计特征。但分子天然是一张"图"——原子是点，化学键是连线，而当时的神经网络（如 CNN、RNN）只擅长处理图片和序列这种规则数据，面对图这种不规则结构就没了办法。这篇论文想让神经网络直接"读懂"分子图，又快又准地预测化学性质。

**💡 主要贡献：** 论文最大的贡献不是发明一个全新模型，而是"统一江湖"：作者发现之前各家提出的多种图神经网络，本质上都在做同一件事——消息传递（Message Passing），于是提出了统一框架 MPNN。就像发现各种方言其实共用一套语法。在此框架下，他们设计的模型在 QM9 分子数据集上达到了化学精度，比之前的方法准得多、比量子计算快几十万倍。

**🔧 方法：** 核心思想像"邻里之间传纸条"：每个原子（节点）先把自己的信息写成"消息"发给相邻的原子；每个原子收到邻居的消息后，更新自己的"认识"；这个过程重复几轮，每个原子就能了解到越来越远的邻居的情况。最后一步叫"读出"（readout），把所有原子的信息汇总成整个分子的表示，用来预测性质。

**🌟 意义：** 这篇论文是现代图神经网络（GNN）的奠基之作。它证明了深度学习不仅能处理图片和文字，还能处理图结构数据——而现实世界中图无处不在：社交网络、推荐系统、交通路网、蛋白质结构。后来 AlphaFold 预测蛋白质结构、药物发现、芯片设计等突破，都能看到消息传递思想的影子。它出现在这份必读清单里，正是因为它把深度学习的疆域扩展到了"关系与结构"的世界。


## 🎯 核心结论 (Key Takeaways)

- **论文的核心发现是"统一"而非"发明"**：Gilmer 等人指出，之前各家提出的多种图神经网络（GCN、门控图网络等）本质上都在做同一件事——消息传递（Message Passing），并据此提出统一框架 MPNN，其模型在 QM9 数据集上达到化学精度，速度比量子力学计算快几个数量级。
- **消息传递只需三步**：本 notebook 用纯 NumPy 实现了这三步——邻居生成消息（message）→ 汇总消息（aggregate）→ 更新自身状态（update），最后再加一个读出（readout）把所有节点求和压成整图预测。在水分子（3 个节点、4 条有向边）上跑一步，就能看到氧原子的状态被两个氢邻居的信息改变。
- **层数 = 信息传播距离**：完整 MPNN 堆叠了 3 层消息传递，节点大约能"听到"3 跳以外的邻居；热力图可视化展示了 4 个快照（初始嵌入 + 3 轮传递）中节点表示逐步演变、不断吸收邻居信息的过程。
- **同一个模型通吃不同结构的图**：训练在水分子（3 节点）上定义的 MPNN，不改一行代码就能直接对苯环 C6H6（12 节点：6 碳成环 + 6 氢外挂）做预测——这正是 GNN 区别于 CNN/RNN 的关键优势：天然适应可变大小、任意形状的图。
- **聚合方式是重要的设计选择**：实验对比了求和（保留总量、对邻居数敏感）、求平均（关注典型值）、取最大（只留最强信号）三种聚合函数，同样的消息会得到完全不同的汇总结果，对应 GCN/GraphSAGE/GIN 等变体的差异来源。
- **带走一句话**：把"原子当节点、化学键当边，让邻居互相传纸条再汇总"这一个简单想法，就把深度学习从图片和序列扩展到了图结构世界——这是现代 GNN（乃至 AlphaFold、药物发现）的奠基思想。


## 🤯 反常识的发现 (Counterintuitive Findings)

- **常识认为：GCN、门控图网络等五花八门的模型是互相竞争的不同发明。** 但这篇论文发现它们本质上是同一个"传消息"框架（MPNN）的变体——区别只在于消息函数、聚合方式和更新函数怎么选。就像发现各地方言其实共用一套语法，论文最大的贡献不是"发明新模型"，而是"统一江湖"。

- **常识认为：每个原子只跟直接相邻的原子交换信息，注定"目光短浅"，看不到全局。** 但实验发现，只要把纯局部的消息传递重复几轮，信息就会像涟漪一样扩散开——本 notebook 里堆 3 层就让节点"听到"3 跳以外的邻居，热力图的 4 个快照清楚地展示了节点表示如何一步步吸收越来越远的信息。局部规则重复几次，就涌现出全局视野。

- **常识认为：神经网络的输入尺寸是固定的（CNN 要固定分辨率、RNN 要固定词表），换个输入结构就得改网络。** 但实验发现，在水分子（3 个节点）上定义的 MPNN，零改动就能直接预测苯环 C6H6（12 个节点、成环结构）——因为网络学的是"节点如何与邻居交流"的规则，而不是某张图的固定形状，所以天然通吃任意大小、任意拓扑的图。

- **常识认为：预测分子性质要么靠昂贵的量子力学计算（一个分子算几小时），要么靠化学家人工设计特征。** 但论文发现，让网络自己在图上"传纸条"学出来的表示，就能在 QM9 上达到化学精度，速度还比量子计算快几个数量级——不懂薛定谔方程的神经网络，靠消息传递也能"算"化学。

## Graph Representation(图的表示)

#### 💻 代码解读

**做什么:** 定义一个简单的图数据结构 `Graph` 类，并用它搭建一个水分子(H2O)的图，然后画出来。

**怎么做:**
- `Graph` 类用三样东西描述一张图: `edges`(边列表，记录谁连着谁)、`node_features`(每个节点的特征向量)、`edge_features`(每条边的特征)。可以把节点想象成"人"，边想象成"朋友关系"。
- `add_edge` 添加一条有向边；`get_neighbors` 找出某个节点的所有邻居；`visualize` 借助 networkx 把图画成节点+箭头的示意图。
- 创建 `water` 图: 3 个节点(1 个氧 O + 2 个氢 H)，氧和每个氢之间来回各加一条边(用两条有向边模拟无向连接)。
- 节点特征是 `[原子序数, 化合价]`，比如氧是 `[8, 2]`、氢是 `[1, 1]`。
- 最后画出水分子的图结构，并打印节点数、边数以及氧原子(节点 0)的邻居。

````python
class Graph:
    """Simple graph representation"""
    def __init__(self, num_nodes):
        self.num_nodes = num_nodes
        # 用边列表表示图结构:每条有向边是一个(源节点,目标节点)元组
        self.edges = []  # List of (source, target) tuples
        self.node_features = []  # List of node feature vectors
        self.edge_features = {}  # Dict: (src, tgt) -> edge features
    
    def add_edge(self, src, tgt, features=None):
        self.edges.append((src, tgt))
        if features is not None:
            self.edge_features[(src, tgt)] = features
    
    def set_node_features(self, features):
        """features: list of feature vectors"""
        self.node_features = features
    
    def get_neighbors(self, node):
        """Get all neighbors of a node"""
        neighbors = []
        # 遍历所有边找出以node为源的边,即出边邻居;元组解包成src和tgt
        for src, tgt in self.edges:
            if src == node:
                neighbors.append(tgt)
        return neighbors
    
    def visualize(self, node_labels=None):
        """Visualize graph using networkx"""
        G = nx.DiGraph()
        G.add_nodes_from(range(self.num_nodes))
        G.add_edges_from(self.edges)
        
        pos = nx.spring_layout(G, seed=42)
        
        plt.figure(figsize=(10, 8))
        nx.draw(G, pos, with_labels=True, node_color='lightblue', 
               node_size=800, font_size=12, arrows=True,
               arrowsize=20, edge_color='gray', width=2)
        
        if node_labels:
            nx.draw_networkx_labels(G, pos, node_labels, font_size=10)
        
        plt.title("Graph Structure")
        plt.axis('off')
        plt.show()

# Create sample molecular graph
# H2O (water): O connected to 2 H atoms
water = Graph(num_nodes=3)
# 无向图用两条方向相反的有向边表示:O->H和H->O各加一次
water.add_edge(0, 1)  # O -> H
water.add_edge(0, 2)  # O -> H  
water.add_edge(1, 0)  # H -> O (undirected)
water.add_edge(2, 0)  # H -> O

# Node features: [atomic_num, valence, ...]
# 每个节点(原子)的初始特征向量:[原子序数,化合价],这是GNN的输入
water.set_node_features([
    np.array([8, 2]),  # Oxygen
    np.array([1, 1]),  # Hydrogen
    np.array([1, 1]),  # Hydrogen
])

labels = {0: 'O', 1: 'H', 2: 'H'}
water.visualize(labels)

print(f"Number of nodes: {water.num_nodes}")
print(f"Number of edges: {len(water.edges)}")
print(f"Neighbors of node 0 (Oxygen): {water.get_neighbors(0)}")
````

## Message Passing Framework(消息传递框架)

**Two phases:**
1. **Message Passing**: Aggregate information from neighbors (T steps)
2. **Readout**: Global graph representation

$$m_v^{t+1} = \sum_{w \in N(v)} M_t(h_v^t, h_w^t, e_{vw})$$
$$h_v^{t+1} = U_t(h_v^t, m_v^{t+1})$$

**两个阶段：**
1. **消息传递(Message Passing)**：从邻居节点聚合信息（共 T 步）
2. **读出(Readout)**：得到全局的图表示

#### 💻 代码解读

**做什么:** 实现消息传递的核心组件 `MessagePassingLayer`(单层消息传递)，并在水分子图上跑一步，看节点状态如何被邻居信息更新。

**怎么做:**
- `message` 函数: 把"发送方状态 + 接收方状态 + 边特征"拼接起来，乘上权重矩阵 `W_msg` 再过 `tanh` 激活，生成一条"消息"——好比邻居给你写了一张便条，内容由你们俩的情况共同决定。
- `aggregate` 函数: 把收到的所有消息按元素求和(`np.sum`)，汇总成一条总消息。
- `update` 函数: 把节点自己当前的状态和汇总消息拼在一起，通过 `W_update` 和 `tanh` 算出节点的新状态——相当于"结合大家的意见更新自己的看法"。
- `forward` 函数: 对图中每个节点 v，收集所有邻居 w 发来的消息 → 聚合 → 更新，返回全部节点的新状态。
- 最后在 `water` 图上测试: 把原始特征补零扩展到 4 维作为初始状态 `initial_states`，跑一次 `mp_layer.forward`，打印氧原子更新前后的状态对比。

````python
class MessagePassingLayer:
    """Single message passing layer"""
    def __init__(self, node_dim, edge_dim, hidden_dim):
        self.node_dim = node_dim
        self.edge_dim = edge_dim
        self.hidden_dim = hidden_dim
        
        # Message function: M(h_v, h_w, e_vw)
        # 消息网络的权重:输入是[源节点,目标节点,边特征]拼接后的向量(2*node_dim+edge_dim维)
        # 乘0.01做小随机初始化,防止tanh一开始就饱和
        self.W_msg = np.random.randn(hidden_dim, 2*node_dim + edge_dim) * 0.01
        self.b_msg = np.zeros(hidden_dim)
        
        # Update function: U(h_v, m_v)
        # 更新网络的权重:输入是[节点当前状态,聚合消息]的拼接,输出回到node_dim维
        self.W_update = np.random.randn(node_dim, node_dim + hidden_dim) * 0.01
        self.b_update = np.zeros(node_dim)
    
    def message(self, h_source, h_target, e_features):
        """Compute message from source to target"""
        # Concatenate source, target, edge features
        if e_features is None:
            e_features = np.zeros(self.edge_dim)
        
        # 形状:(node_dim,)+(node_dim,)+(edge_dim,) -> (2*node_dim+edge_dim,)
        concat = np.concatenate([h_source, h_target, e_features])
        
        # Apply message network
        # 单层感知机生成消息:对应论文中的消息函数M_t,输出形状(hidden_dim,)
        message = np.tanh(np.dot(self.W_msg, concat) + self.b_msg)
        return message
    
    def aggregate(self, messages):
        """Aggregate messages (sum)"""
        # 孤立节点没有邻居消息,返回零向量避免空列表求和报错
        if len(messages) == 0:
            return np.zeros(self.hidden_dim)
        # 沿axis=0把所有邻居的消息逐元素求和:求和对邻居顺序不变(置换不变性),这是GNN的关键性质
        return np.sum(messages, axis=0)
    
    def update(self, h_node, aggregated_message):
        """Update node representation"""
        # 拼接旧状态与聚合消息,形状:(node_dim+hidden_dim,)
        concat = np.concatenate([h_node, aggregated_message])
        # 对应论文的更新函数U_t:结合自身旧状态和邻居信息得到新状态,形状(node_dim,)
        h_new = np.tanh(np.dot(self.W_update, concat) + self.b_update)
        return h_new
    
    def forward(self, graph, node_states):
        """
        One message passing step
        
        graph: Graph object
        node_states: list of current node hidden states
        
        Returns: updated node states
        """
        new_states = []
        
        # 对每个节点执行"收消息->聚合->更新"三步,即一轮完整的消息传递
        for v in range(graph.num_nodes):
            # Collect messages from neighbors
            messages = []
            for w in graph.get_neighbors(v):
                # Get edge features
                # dict.get第二个参数是默认值:边(w,v)没有特征时返回None
                edge_feat = graph.edge_features.get((w, v), None)
                
                # Compute message
                # 消息由邻居w发往节点v,携带w的状态、v的状态和边特征
                msg = self.message(node_states[w], node_states[v], edge_feat)
                messages.append(msg)
            
            # Aggregate messages
            aggregated = self.aggregate(messages)
            
            # Update node state
            h_new = self.update(node_states[v], aggregated)
            new_states.append(h_new)
        
        return new_states

# Test message passing
node_dim = 4
edge_dim = 2
hidden_dim = 8

mp_layer = MessagePassingLayer(node_dim, edge_dim, hidden_dim)

# Initialize node states from features
initial_states = []
for feat in water.node_features:
    # Embed to higher dimension
    # 零填充把2维原始特征扩到node_dim维,作为初始隐藏状态h_v^0
    state = np.concatenate([feat, np.zeros(node_dim - len(feat))])
    initial_states.append(state)

# Run message passing
updated_states = mp_layer.forward(water, initial_states)

print(f"\nInitial state (O): {initial_states[0]}")
print(f"Updated state (O): {updated_states[0]}")
print(f"\nNode states updated via neighbor information!")
````

## Complete MPNN(完整的 MPNN)

#### 💻 代码解读

**做什么:** 把多层消息传递组装成完整的 `MPNN` 网络(嵌入 → 多轮消息传递 → 读出)，对整个水分子做一次图级别的性质预测。

**怎么做:**
- `MPNN` 类由三部分组成: 嵌入层 `embed_W`(把 2 维原子特征映射到 8 维隐藏空间)、3 个堆叠的 `MessagePassingLayer`(消息传递层)、读出层 `readout_W`(把整张图压缩成一个预测值)。
- `forward` 流程: 先把每个节点特征嵌入成隐藏状态；然后逐层做消息传递，每传一轮，节点能"听到"更远邻居的信息(3 层约等于信息能传 3 跳)；同时用 `states_history` 记录每一步的节点状态，供后面可视化。
- 读出(Readout)阶段: 把所有节点的最终状态简单求和(sum pooling)得到整图表示 `graph_repr`，再做一次线性变换输出预测值——好比把每个成员的意见汇总成整个团队的结论。
- 最后创建 `mpnn` 实例，在 `water` 图上前向传播，打印图级别预测值(可类比分子能量、溶解度等性质)。

````python
class MPNN:
    """Message Passing Neural Network"""
    def __init__(self, node_feat_dim, edge_feat_dim, hidden_dim, num_layers, output_dim):
        self.hidden_dim = hidden_dim
        self.num_layers = num_layers
        
        # Embedding layer
        self.embed_W = np.random.randn(hidden_dim, node_feat_dim) * 0.01
        
        # Message passing layers
        # 列表推导式创建num_layers个独立的消息传递层;层数T决定信息能传播多远(T跳邻居)
        self.mp_layers = [
            MessagePassingLayer(hidden_dim, edge_feat_dim, hidden_dim*2)
            for _ in range(num_layers)
        ]
        
        # Readout (graph-level prediction)
        self.readout_W = np.random.randn(output_dim, hidden_dim) * 0.01
        self.readout_b = np.zeros(output_dim)
    
    def forward(self, graph):
        """
        Forward pass through MPNN
        
        Returns: graph-level prediction
        """
        # Embed node features
        node_states = []
        for feat in graph.node_features:
            # 线性映射+tanh把原始特征嵌入到hidden_dim维,形状:(node_feat_dim,)->(hidden_dim,)
            embedded = np.tanh(np.dot(self.embed_W, feat))
            node_states.append(embedded)
        
        # Message passing
        # 记录每一步的节点状态,便于后面可视化表示如何演化
        states_history = [node_states]
        # 逐层消息传递:第t层后每个节点已聚合了t跳范围内的信息
        for layer in self.mp_layers:
            node_states = layer.forward(graph, node_states)
            states_history.append(node_states)
        
        # Readout: aggregate node states to graph representation
        # 读出函数R:把所有节点状态求和成一个图级向量,对节点顺序不变,且适配任意节点数
        graph_repr = np.sum(node_states, axis=0)  # Simple sum pooling
        
        # Final prediction
        # 线性层把图表示映射为标量属性预测,形状:(hidden_dim,)->(output_dim,)
        output = np.dot(self.readout_W, graph_repr) + self.readout_b
        
        return output, states_history

# Create MPNN
mpnn = MPNN(
    node_feat_dim=2,
    edge_feat_dim=2,
    hidden_dim=8,
    num_layers=3,
    output_dim=1  # Predict single property (e.g., energy)
)

# Forward pass
prediction, history = mpnn.forward(water)

print(f"Graph-level prediction: {prediction}")
print(f"(E.g., molecular property like energy, solubility, etc.)")
````

## Visualize Message Passing(可视化消息传递)

#### 💻 代码解读

**做什么:** 用热力图可视化消息传递过程中节点表示(隐藏状态)是如何一步步演变的。

**怎么做:**
- 遍历上一步保存的 `history`(共 4 个快照: 初始嵌入 + 3 层消息传递后的状态)，每个快照画一张子图。
- 把节点状态堆成矩阵 `states_matrix`(行=8 个隐藏维度，列=O、H、H 三个节点)，用 `imshow` 画成红蓝热力图——颜色深浅代表激活值大小。
- 加上颜色条和标题后展示: 可以直观看到随着传递步数增加，各节点的表示逐渐变化，说明它们在不断吸收邻居的信息。

````python
# Visualize how node representations evolve
fig, axes = plt.subplots(1, len(history), figsize=(16, 4))

for step, states in enumerate(history):
    # Stack node states for visualization
    # 把节点状态列表堆成矩阵再转置,形状:(num_nodes, hidden_dim)->(hidden_dim, num_nodes)
    states_matrix = np.array(states).T  # (hidden_dim, num_nodes)
    
    ax = axes[step]
    # 用热力图展示每个节点的隐藏向量;红蓝配色便于看正负激活
    im = ax.imshow(states_matrix, cmap='RdBu', aspect='auto')
    ax.set_title(f'Step {step}')
    ax.set_xlabel('Node')
    ax.set_ylabel('Hidden Dimension')
    ax.set_xticks([0, 1, 2])
    ax.set_xticklabels(['O', 'H', 'H'])

plt.colorbar(im, ax=axes, label='Activation')
plt.suptitle('Node Representations Through Message Passing', fontsize=14)
plt.tight_layout()
plt.show()

print("\nNodes update their representations by aggregating neighbor information")
````

## Create More Complex Graph(构建更复杂的图)

#### 💻 代码解读

**做什么:** 构建一个更复杂的分子——苯环(C6H6)的图，画出来并用刚才的 MPNN 做预测，验证同一个模型可以处理任意结构的图。

**怎么做:**
- 创建 12 个节点的 `benzene` 图: 节点 0-5 是碳(C)，节点 6-11 是氢(H)。
- 第一个循环用 `(i + 1) % 6` 把 6 个碳首尾相连成一个环(像 6 个人手拉手围成圈)，每对之间加双向边。
- 第二个循环把每个氢挂到对应的碳上(碳 i 连氢 6+i)，同样是双向边。
- 设置节点特征: 碳是 `[6, 3]`，氢是 `[1, 1]`(原子序数和化合价)。
- 调用 `visualize` 画出苯环结构，再把图喂给同一个 `mpnn.forward`——不需要改任何代码，因为 MPNN 天生适应不同大小、不同形状的图——最后打印苯的预测值。

````python
# Create benzene ring (C6H6)
benzene = Graph(num_nodes=12)  # 6 C + 6 H

# Carbon ring (nodes 0-5)
for i in range(6):
    # 取模让第5号碳连回第0号,形成六元环
    next_i = (i + 1) % 6
    benzene.add_edge(i, next_i)
    benzene.add_edge(next_i, i)

# Hydrogen atoms (nodes 6-11) attached to carbons
for i in range(6):
    h_idx = 6 + i
    benzene.add_edge(i, h_idx)
    benzene.add_edge(h_idx, i)

# Node features
features = []
for i in range(6):
    features.append(np.array([6, 3]))  # Carbon
for i in range(6):
    features.append(np.array([1, 1]))  # Hydrogen
benzene.set_node_features(features)

# Visualize
labels = {i: 'C' for i in range(6)}
labels.update({i: 'H' for i in range(6, 12)})
benzene.visualize(labels)

# Run MPNN
# 同一个MPNN不需任何改动就能处理不同大小的图(3节点的水->12节点的苯):这是GNN相对MLP/CNN的核心优势
pred_benzene, hist_benzene = mpnn.forward(benzene)
print(f"\nBenzene prediction: {pred_benzene}")
````

## Different Aggregation Functions(不同的聚合函数)

#### 💻 代码解读

**做什么:** 对比三种消息聚合(aggregation)策略——求和、求平均、取最大——展示不同聚合方式会得到不同的汇总结果。

**怎么做:**
- 定义三个聚合函数: `sum_aggregation`(逐元素求和，保留"总量"信息，对邻居数量敏感)、`mean_aggregation`(求平均，关注"典型值"，不受邻居多少影响)、`max_aggregation`(取最大，只保留最突出的信号)。可以类比统计一个班的成绩: 总分、平均分、最高分各说明不同问题。
- 生成 3 条 8 维的随机消息 `test_messages` 作为测试输入。
- 分别用三种方式聚合并打印前 4 个数值，直观对比: 同样的消息，不同聚合函数抓住的模式不一样，这也是 GNN 设计中的一个重要选择。

````python
# Compare aggregation strategies
# 三种聚合都满足置换不变性,但表达能力不同:sum保留邻居数量信息(区分度最强)
def sum_aggregation(messages):
    return np.sum(messages, axis=0) if len(messages) > 0 else np.zeros_like(messages[0])

# mean对节点度数不敏感,适合邻居数差异很大的图
def mean_aggregation(messages):
    return np.mean(messages, axis=0) if len(messages) > 0 else np.zeros_like(messages[0])

# max逐维取最大值,只关注"最显著"的邻居信号
def max_aggregation(messages):
    return np.max(messages, axis=0) if len(messages) > 0 else np.zeros_like(messages[0])

# Test on random messages
test_messages = [np.random.randn(8) for _ in range(3)]

print("Aggregation Functions:")
print(f"Sum: {sum_aggregation(test_messages)[:4]}...")
print(f"Mean: {mean_aggregation(test_messages)[:4]}...")
print(f"Max: {max_aggregation(test_messages)[:4]}...")
print("\nDifferent aggregations capture different patterns!")
````

## Key Takeaways(要点总结)

### Message Passing Framework:

**Phase 1: Message Passing** (repeat T times)
```
For each node v:
  1. Collect messages from neighbors:
     m_v = Σ_{u∈N(v)} M_t(h_v, h_u, e_uv)
  
  2. Update node state:
     h_v = U_t(h_v, m_v)
```

**Phase 2: Readout**
```
Graph representation:
  h_G = R({h_v | v ∈ G})
```

### Components:
1. **Message function M**: Compute message from neighbor
2. **Aggregation**: Combine messages (sum, mean, max, attention)
3. **Update function U**: Update node representation
4. **Readout R**: Graph-level pooling

### Variants:
- **GCN**: Simplified message passing with normalization
- **GraphSAGE**: Sampling neighbors, inductive learning
- **GAT**: Attention-based aggregation
- **GIN**: Powerful aggregation (sum + MLP)

### Applications:
- **Molecular property prediction**: QM9, drug discovery
- **Social networks**: Node classification, link prediction
- **Knowledge graphs**: Reasoning, completion
- **Recommendation**: User-item graphs
- **3D vision**: Point clouds, meshes

### Advantages:
- ✅ Handles variable-size graphs
- ✅ Permutation invariant
- ✅ Inductive learning (generalize to new graphs)
- ✅ Interpretable (message passing)

### Challenges:
- Over-smoothing (deep layers make nodes similar)
- Expressiveness (limited by aggregation)
- Scalability (large graphs)

### Modern Extensions:
- **Graph Transformers**: Attention on full graph
- **Equivariant GNNs**: Respect symmetries (E(3), SE(3))
- **Temporal GNNs**: Dynamic graphs
- **Heterogeneous GNNs**: Multiple node/edge types

### 消息传递框架：

**阶段 1：消息传递**（重复 T 次）——对每个节点 v：先从邻居收集消息 `m_v = Σ_{u∈N(v)} M_t(h_v, h_u, e_uv)`，再更新节点状态 `h_v = U_t(h_v, m_v)`。

**阶段 2：读出(Readout)**——得到图表示 `h_G = R({h_v | v ∈ G})`。

### 组成部分：
1. **消息函数 M**：计算来自邻居的消息
2. **聚合(Aggregation)**：组合消息（求和 sum、平均 mean、最大值 max、注意力 attention）
3. **更新函数 U**：更新节点表示
4. **读出 R**：图级别的池化(pooling)

### 变体：
- **GCN**：带归一化的简化消息传递
- **GraphSAGE**：邻居采样，归纳式学习(inductive learning)
- **GAT**：基于注意力(attention)的聚合
- **GIN**：更强表达力的聚合（求和 + MLP）

### 应用：
- **分子性质预测**：QM9、药物发现
- **社交网络**：节点分类、链接预测(link prediction)
- **知识图谱**：推理、补全
- **推荐系统**：用户-物品图
- **3D 视觉**：点云(point clouds)、网格(meshes)

### 优点：
- ✅ 能处理可变大小的图
- ✅ 置换不变性(permutation invariant)
- ✅ 归纳式学习（可泛化到新的图）
- ✅ 可解释性强（消息传递机制）

### 挑战：
- 过平滑(over-smoothing)（层数加深会使节点表示趋同）
- 表达能力（受聚合函数限制）
- 可扩展性（大规模图）

### 现代扩展：
- **图 Transformer(Graph Transformers)**：在全图上做注意力
- **等变 GNN(Equivariant GNNs)**：遵循对称性（E(3)、SE(3)）
- **时序 GNN(Temporal GNNs)**：动态图
- **异构 GNN(Heterogeneous GNNs)**：多种节点/边类型
