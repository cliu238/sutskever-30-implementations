## 📖 论文导读

**🎯 这篇文章想解决什么问题（目的）：** 让计算机学会"看懂"照片——给它一张图，它能说出里面是猫、狗还是汽车。在 2012 年之前，图像识别主要靠人工设计特征（相当于人先告诉机器"看什么"），效果进步缓慢；ImageNet 比赛有 120 万张图片、1000 个类别，传统方法的错误率一直居高不下。这篇论文想证明：只要有足够的数据和算力，让一个深层神经网络自己从原始像素中学习特征，就能大幅超越人工设计的方法。

**💡 主要贡献：** 训练出了当时最大的卷积神经网络 AlexNet（5 层卷积 + 3 层全连接，约 6000 万参数），在 ImageNet 2012 比赛中把 top-5 错误率从第二名的 26.2% 一举降到 15.3%，优势之大震惊了整个领域。它还带火了几项后来成为标配的技术：ReLU 激活函数（让训练快好几倍）、Dropout（随机"关掉"部分神经元防止死记硬背）、数据增强（把图片裁剪翻转来"变出"更多训练样本），以及用 GPU 加速训练。

**🔧 方法：** 卷积网络的思路可以类比人眼看东西：先看局部的边缘和纹理，再逐层组合成形状、部件，最后认出整个物体。AlexNet 用小窗口（卷积核）在图片上滑动扫描提取局部特征，层层堆叠加深理解；训练时用两块 GPU 并行计算，把原本要几个月的训练缩短到几天。整个过程端到端：输入原始像素，输出类别，中间的特征全部由网络自己学出来，不需要人工设计。

**🌟 意义：** 这篇论文被公认为深度学习革命的"引爆点"。它用无可辩驳的成绩证明了"深度网络 + 大数据 + GPU"这条路走得通，此后计算机视觉乃至整个 AI 领域全面转向深度学习。今天的图像识别、自动驾驶、医学影像分析都是沿着它开辟的道路发展而来。对本阅读清单而言，它也是 Ilya Sutskever（本清单的推荐者之一）的成名作——理解 AlexNet，就理解了现代深度学习浪潮的起点。

## 🤯 反常识的发现 (Counterintuitive Findings)

- **常识认为激活函数应该平滑优雅**——sigmoid 和 tanh 曲线圆润、处处可导，数学上"漂亮"得多。但这篇论文发现，最"粗糙"的 ReLU（本 notebook 里就一行 `np.maximum(0, x)`，一条折线）反而让训练快了约 6 倍：正是因为它简单粗暴、正区间梯度恒为 1，信号才不会像 sigmoid 那样在深层网络里被"压扁"到消失。越笨的函数，跑得越快。

- **常识认为让机器"带病工作"只会更差**——谁会故意在训练时随机"杀死"一半神经元？但 Dropout 恰恰这么干（丢弃率 0.5），结果 6000 万参数的网络反而不再死记硬背：每个神经元都不能指望"队友"一定在场，只能各自学到真本事，相当于免费训练了无数个子网络的集成。

- **常识认为篡改训练数据是作弊**——把图片翻转、裁剪、加噪声后再喂给模型，听起来像在污染数据。但 AlexNet 靠这招把训练集"免费"扩大了上千倍（本 notebook 的数据增强实验把 1 张图变出 7 个不同版本），网络见过的"猫"越歪越花，认真实的猫反而越准。

- **常识认为精心设计的特征工程必然胜过"暴力"学习**——2012 年之前，主流视神经网络为死路，大家都在人工设计 SIFT、HOG 等特征。但两块打游戏用的 GTX 580 显卡 + 一个自己从像素学特征的深网，把 top-5 错误率从 26.2% 干到 15.3%，领先第二名近 11 个百分点。本 notebook 的滤波器可视化也印证了这一点：没人告诉网络"看什么"，第一层却自动长出了边缘和颜色检测器。

## Convolutional Layer Implementation(卷积层实现)

The core building block of CNNs

卷积神经网络(CNN)的核心构建模块

#### 💻 代码解读

**做什么:** 用纯 NumPy 从零实现 CNN 的三个核心零件——ReLU 激活、二维卷积和最大池化，并做一次小测试。

**怎么做:**
- 定义 `relu(x)`：用 `np.maximum(0, x)` 把负数全部变成 0，正数原样保留，相当于一个"只放行正信号"的闸门；
- 定义 `conv2d(input_image, kernel, stride, padding)`：先给图像四周补零（padding），再让卷积核像"手电筒扫描"一样在图上滑动，每停一处就取出一小块 `patch`，与卷积核逐元素相乘再求和，得到输出的一个像素；
- 定义 `max_pool2d(input_image, pool_size, stride)`：把特征图切成 2x2 的小格子，每格只保留最大值（`np.max`），相当于把图片"缩小"但保留最显眼的特征；
- 最后用一张 8x8 的随机图 `test_image` 和 3 个 3x3 卷积核 `test_kernel` 测试：打印卷积后形状 (3, 8, 8) 和池化后形状 (3, 4, 4)，验证代码正确。

````python
def relu(x):
    return np.maximum(0, x)

def conv2d(input_image, kernel, stride=1, padding=0):
    """
    2D Convolution operation
    
    input_image: (H, W) or (C, H, W)
    kernel: (out_channels, in_channels, kH, kW)
    """
    if len(input_image.shape) == 2:
        # np.newaxis在最前面插入通道维:形状(H, W) -> (1, H, W),统一按多通道处理
        input_image = input_image[np.newaxis, :, :]
    
    in_channels, H, W = input_image.shape
    out_channels, _, kH, kW = kernel.shape
    
    # Add padding
    if padding > 0:
        # 只在H、W两个空间维度四周补零,通道维(0,0)不补;padding=1可让3x3卷积保持尺寸不变
        input_padded = np.pad(input_image, 
                             ((0, 0), (padding, padding), (padding, padding)), 
                             mode='constant')
    else:
        input_padded = input_image
    
    # Output dimensions
    # 卷积输出尺寸公式:(输入 + 2*padding - 核大小) // 步长 + 1
    out_H = (H + 2*padding - kH) // stride + 1
    out_W = (W + 2*padding - kW) // stride + 1
    
    output = np.zeros((out_channels, out_H, out_W))
    
    # Perform convolution
    for oc in range(out_channels):
        for i in range(out_H):
            for j in range(out_W):
                h_start = i * stride
                w_start = j * stride
                
                # Extract patch
                # 滑动窗口:切出与卷积核同大的局部块,形状(in_channels, kH, kW)
                patch = input_padded[:, h_start:h_start+kH, w_start:w_start+kW]
                
                # Convolve with kernel
                # 逐元素相乘再求和(跨所有输入通道),即局部块与核的点积——卷积的本质
                output[oc, i, j] = np.sum(patch * kernel[oc])
    
    return output

def max_pool2d(input_image, pool_size=2, stride=2):
    """
    Max pooling operation
    """
    C, H, W = input_image.shape
    
    out_H = (H - pool_size) // stride + 1
    out_W = (W - pool_size) // stride + 1
    
    output = np.zeros((C, out_H, out_W))
    
    for c in range(C):
        for i in range(out_H):
            for j in range(out_W):
                h_start = i * stride
                w_start = j * stride
                
                pool_region = input_image[c, h_start:h_start+pool_size, 
                                         w_start:w_start+pool_size]
                # 取窗口内最大值:下采样的同时保留最强响应,带来轻微的平移不变性
                output[c, i, j] = np.max(pool_region)
    
    return output

# Test convolution
test_image = np.random.randn(1, 8, 8)
test_kernel = np.random.randn(3, 1, 3, 3) * 0.1

conv_output = conv2d(test_image, test_kernel, stride=1, padding=1)
print(f"Input shape: {test_image.shape}")
print(f"Kernel shape: {test_kernel.shape}")
print(f"Conv output shape: {conv_output.shape}")

pooled = max_pool2d(conv_output, pool_size=2, stride=2)
print(f"After max pooling: {pooled.shape}")
````

## AlexNet Architecture (Simplified)(AlexNet 架构（简化版）)

Original: 227x227x3 → 5 conv layers → 3 FC layers → 1000 classes

Our simplified version for 32x32 images

原版：227x227x3 → 5 个卷积层(conv layers) → 3 个全连接层(FC layers) → 1000 个类别

我们针对 32x32 图像的简化版本

#### 💻 代码解读

**做什么:** 定义一个简化版的 AlexNet 类 `AlexNetSimplified`，适配 32x32 小图（类似 CIFAR-10），并跑一次前向传播测试。

**怎么做:**
- `__init__` 中用小随机数初始化所有参数：三层卷积核 `conv1_filters`（32 个）、`conv2_filters`（64 个）、`conv3_filters`（128 个），以及两层全连接权重 `fc1_weights`（2048→512）和 `fc2_weights`（512→10）——就像盖楼前先备好三层"特征提取车间"和两层"决策办公室"的材料；
- `forward` 方法串起完整流水线：每层都是"卷积 → 加偏置 → ReLU → 最大池化"，图像边长从 32 逐步缩到 16、8、4，通道数从 3 增加到 128；
- 卷积结束后用 `reshape(-1)` 把 128x4x4 的立体特征"压扁"成 2048 维向量，送入全连接层；
- FC1 之后可选开启 dropout（随机屏蔽一半神经元并按比例放大剩余值，防止网络死记硬背），FC2 输出 10 个类别分数；
- 最后创建模型实例 `alexnet`，用一张随机图测试，打印输出形状 (10,)，即 10 个类别的得分。

````python
class AlexNetSimplified:
    def __init__(self, num_classes=10):
        """
        Simplified AlexNet for 32x32 images (like CIFAR-10)
        
        Architecture:
        - Conv1: 3x3x3 -> 32 filters
        - MaxPool
        - Conv2: 32 -> 64 filters
        - MaxPool
        - Conv3: 64 -> 128 filters
        - FC layers
        """
        # Conv layers
        # 权重形状(out_channels, in_channels, kH, kW);乘0.01保持初始值很小,防止激活爆炸
        self.conv1_filters = np.random.randn(32, 3, 3, 3) * 0.01
        self.conv1_bias = np.zeros(32)
        
        self.conv2_filters = np.random.randn(64, 32, 3, 3) * 0.01
        self.conv2_bias = np.zeros(64)
        
        self.conv3_filters = np.random.randn(128, 64, 3, 3) * 0.01
        self.conv3_bias = np.zeros(128)
        
        # FC layers (after conv: 128 * 4 * 4 = 2048)
        # 卷积输出128x4x4展平成2048维向量,再接全连接层
        self.fc1_weights = np.random.randn(2048, 512) * 0.01
        self.fc1_bias = np.zeros(512)
        
        self.fc2_weights = np.random.randn(512, num_classes) * 0.01
        self.fc2_bias = np.zeros(num_classes)
    
    def forward(self, x, use_dropout=False, dropout_rate=0.5):
        """
        Forward pass
        x: (3, 32, 32) image
        """
        # Conv1 + ReLU + MaxPool
        conv1 = conv2d(x, self.conv1_filters, stride=1, padding=1)
        # 广播加偏置:(32,) -> (32,1,1),自动扩展到每个空间位置;形状(32,32,32)
        conv1 += self.conv1_bias[:, np.newaxis, np.newaxis]
        conv1 = relu(conv1)
        # 2x2最大池化使空间尺寸减半:(32,32,32) -> (32,16,16)
        pool1 = max_pool2d(conv1, pool_size=2, stride=2)  # 32 x 16 x 16
        
        # Conv2 + ReLU + MaxPool
        conv2 = conv2d(pool1, self.conv2_filters, stride=1, padding=1)
        conv2 += self.conv2_bias[:, np.newaxis, np.newaxis]
        conv2 = relu(conv2)
        pool2 = max_pool2d(conv2, pool_size=2, stride=2)  # 64 x 8 x 8
        
        # Conv3 + ReLU + MaxPool
        conv3 = conv2d(pool2, self.conv3_filters, stride=1, padding=1)
        conv3 += self.conv3_bias[:, np.newaxis, np.newaxis]
        conv3 = relu(conv3)
        pool3 = max_pool2d(conv3, pool_size=2, stride=2)  # 128 x 4 x 4
        
        # Flatten
        # reshape(-1)把(128,4,4)拉平成一维向量(2048,),供全连接层使用
        flattened = pool3.reshape(-1)
        
        # FC1 + ReLU + Dropout
        fc1 = np.dot(flattened, self.fc1_weights) + self.fc1_bias
        fc1 = relu(fc1)
        
        if use_dropout:
            # Dropout(AlexNet的关键正则化):随机把约一半神经元置零,防止共适应/过拟合
            dropout_mask = (np.random.rand(*fc1.shape) > dropout_rate).astype(float)
            # inverted dropout:除以保留概率(1-p),使激活期望不变,测试时无需缩放
            fc1 = fc1 * dropout_mask / (1 - dropout_rate)
        
        # FC2 (output)
        output = np.dot(fc1, self.fc2_weights) + self.fc2_bias
        
        return output

# Create model
alexnet = AlexNetSimplified(num_classes=10)
print("AlexNet (simplified) created")

# Test forward pass
test_img = np.random.randn(3, 32, 32)
output = alexnet.forward(test_img)
print(f"Input: (3, 32, 32)")
print(f"Output: {output.shape} (class scores)")
````

## Generate Synthetic Image Data(生成合成图像数据)

#### 💻 代码解读

**做什么:** 用代码"画"出 10 类简单图案的合成数据集（横条纹、圆形、方块等），代替真实的 ImageNet 图片来做实验，并展示每类样本。

**怎么做:**
- 定义 `generate_simple_images(num_samples, image_size)`：循环生成图片，按 `i % 10` 轮流分配 10 个类别标签；
- 每个类别用不同的几何规则填充像素：比如类别 0 每隔 4 行画 2 行亮条（横条纹）、类别 4 用圆的方程 `(x-center)² + (y-center)² <= radius²` 画圆、类别 3 用棋盘格规则交替填色等；
- 给每张图乘上随机颜色 `color`、加上高斯噪声，再用 `np.clip` 把像素值限制在 [0,1]——让同类图片"长得像但不完全一样"，更接近真实数据；
- 调用函数生成 200 张训练图 `X_train` 和 50 张测试图 `X_test`，并定义类别名列表 `class_names`；
- 最后用 matplotlib 画出 2x5 的网格，每格展示一个类别的样本图及其名称（如 'Circle'、'Cross'）。

````python
def generate_simple_images(num_samples=100, image_size=32):
    """
    Generate simple synthetic images with different patterns
    Classes:
    0: Horizontal stripes
    1: Vertical stripes
    2: Diagonal stripes
    3: Checkerboard
    4: Circle
    5: Square
    6: Cross
    7: Triangle
    8: Random noise
    9: Solid color
    """
    X = []
    y = []
    
    for i in range(num_samples):
        # 取模让10个类别循环出现,保证各类样本数均衡
        class_label = i % 10
        img = np.zeros((3, image_size, image_size))
        
        if class_label == 0:  # Horizontal stripes
            for row in range(0, image_size, 4):
                img[:, row:row+2, :] = 1
        
        elif class_label == 1:  # Vertical stripes
            for col in range(0, image_size, 4):
                img[:, :, col:col+2] = 1
        
        elif class_label == 2:  # Diagonal
            for i in range(image_size):
                if i < image_size:
                    img[:, i, i] = 1
                    if i+1 < image_size:
                        img[:, i, i+1] = 1
        
        elif class_label == 3:  # Checkerboard
            for i in range(0, image_size, 4):
                for j in range(0, image_size, 4):
                    if (i//4 + j//4) % 2 == 0:
                        img[:, i:i+4, j:j+4] = 1
        
        elif class_label == 4:  # Circle
            center = image_size // 2
            radius = image_size // 3
            # np.ogrid生成可广播的坐标网格:y_grid形状(32,1),x_grid形状(1,32)
            y_grid, x_grid = np.ogrid[:image_size, :image_size]
            # 广播计算每个像素到圆心的距离,得到(32,32)的布尔圆形掩码
            mask = (x_grid - center)**2 + (y_grid - center)**2 <= radius**2
            # 布尔掩码索引:一次性把圆内所有像素(全部3个通道)置1
            img[:, mask] = 1
        
        elif class_label == 5:  # Square
            margin = image_size // 4
            img[:, margin:-margin, margin:-margin] = 1
        
        elif class_label == 6:  # Cross
            mid = image_size // 2
            thickness = 3
            img[:, mid-thickness:mid+thickness, :] = 1
            img[:, :, mid-thickness:mid+thickness] = 1
        
        elif class_label == 7:  # Triangle
            for i in range(image_size):
                width = int((i / image_size) * image_size / 2)
                start = image_size // 2 - width
                end = image_size // 2 + width
                img[:, i, start:end] = 1
        
        elif class_label == 8:  # Random noise
            img = np.random.rand(3, image_size, image_size)
        
        else:  # Solid
            img[:] = 0.7
        
        # Add color variation
        # 形状(3,1,1)的随机颜色因子,广播到(3,H,W)给每个通道整体染色
        color = np.random.rand(3, 1, 1)
        img = img * color
        
        # Add noise
        img += np.random.randn(3, image_size, image_size) * 0.1
        # 加噪后裁剪回[0,1],保证仍是合法的像素值
        img = np.clip(img, 0, 1)
        
        X.append(img)
        y.append(class_label)
    
    return np.array(X), np.array(y)

# Generate dataset
X_train, y_train = generate_simple_images(200)
X_test, y_test = generate_simple_images(50)

print(f"Training set: {X_train.shape}")
print(f"Test set: {X_test.shape}")

# Visualize samples
class_names = ['H-Stripes', 'V-Stripes', 'Diagonal', 'Checker', 'Circle', 
               'Square', 'Cross', 'Triangle', 'Noise', 'Solid']

fig, axes = plt.subplots(2, 5, figsize=(15, 6))
axes = axes.flatten()

for i in range(10):
    # Find first occurrence of each class
    # np.where返回满足条件的下标数组,[0][0]取该类第一个样本
    idx = np.where(y_train == i)[0][0]
    # matplotlib要求通道在最后:(C,H,W) -> (H,W,C)
    img = X_train[idx].transpose(1, 2, 0)  # CHW -> HWC
    axes[i].imshow(img)
    axes[i].set_title(class_names[i])
    axes[i].axis('off')

plt.suptitle('Synthetic Image Dataset (10 Classes)', fontsize=14)
plt.tight_layout()
plt.show()
````

## Data Augmentation(数据增强)

AlexNet used data augmentation extensively - a key innovation

AlexNet 大量使用了数据增强(data augmentation)——这是一项关键创新

#### 💻 代码解读

**做什么:** 实现 AlexNet 的关键技巧——数据增强（data augmentation），通过随机翻转、裁剪、加噪声把一张图"变出"多张不同版本，并可视化效果。

**怎么做:**
- `random_flip(img)`：掷硬币决定是否左右镜像翻转（`img[:, :, ::-1]`）——猫的照片翻转后还是猫，但对网络来说就是一张"新"样本；
- `random_crop(img, crop_size=28)`：随机选位置裁出 28x28 的小块，再用最近邻插值放大回 32x32，模拟"拍照时取景稍有偏移"；
- `add_noise(img)`：加上小幅高斯噪声再 `np.clip` 截断，模拟照片的颗粒感；
- `augment_image(img)` 把上面三招串起来一次性施加；
- 最后取训练集第一张图 `X_train[0]`，画出 1 张原图 + 7 张增强后的版本，直观展示同一张图能变出多少花样。

````python
def random_flip(img):
    """Horizontal flip"""
    if np.random.rand() > 0.5:
        # 步长为-1的切片沿宽度轴反转,实现水平翻转;copy()避免返回负步长视图
        return img[:, :, ::-1].copy()
    return img

def random_crop(img, crop_size=28):
    """Random crop"""
    _, h, w = img.shape
    # 随机选裁剪框左上角,对应AlexNet论文中的随机裁剪增广
    top = np.random.randint(0, h - crop_size + 1)
    left = np.random.randint(0, w - crop_size + 1)
    
    cropped = img[:, top:top+crop_size, left:left+crop_size]
    
    # Resize back to original
    # Simple nearest neighbor (for demo)
    scale_h = h / crop_size
    scale_w = w / crop_size
    
    # 最近邻插值把裁剪块拉回原尺寸:每个目标像素取最近的源像素
    resized = np.zeros_like(img)
    for i in range(h):
        for j in range(w):
            src_i = min(int(i / scale_h), crop_size - 1)
            src_j = min(int(j / scale_w), crop_size - 1)
            resized[:, i, j] = cropped[:, src_i, src_j]
    
    return resized

def add_noise(img, noise_level=0.05):
    """Add Gaussian noise"""
    # *img.shape把形状元组解包成多个参数传给randn
    noise = np.random.randn(*img.shape) * noise_level
    return np.clip(img + noise, 0, 1)

def augment_image(img):
    """Apply random augmentations"""
    # 依次叠加翻转+裁剪+噪声:同一张图产生无数变体,等效扩大数据集
    img = random_flip(img)
    img = random_crop(img)
    img = add_noise(img)
    return img

# Demonstrate augmentation
original = X_train[0]

fig, axes = plt.subplots(2, 4, figsize=(16, 8))

axes[0, 0].imshow(original.transpose(1, 2, 0))
axes[0, 0].set_title('Original')
axes[0, 0].axis('off')

for i in range(1, 8):
    augmented = augment_image(original.copy())
    row = i // 4
    col = i % 4
    axes[row, col].imshow(augmented.transpose(1, 2, 0))
    axes[row, col].set_title(f'Augmented {i}')
    axes[row, col].axis('off')

plt.suptitle('Data Augmentation Examples', fontsize=14)
plt.tight_layout()
plt.show()
````

## Visualize Learned Filters(可视化学习到的滤波器)

One of the insights from AlexNet: visualize what the network learns

AlexNet 的洞见之一：将网络学到的内容可视化

#### 💻 代码解读

**做什么:** 把 AlexNet 第一层卷积的 32 个滤波器（filter）画出来看——呼应 AlexNet 论文的著名发现：第一层会自动学出边缘和颜色检测器。

**怎么做:**
- 取出 `alexnet.conv1_filters`，形状为 (32, 3, 3, 3)，即 32 个 3x3 的 RGB 小滤波器；
- 对每个滤波器：先用 `transpose(1, 2, 0)` 把通道顺序从 CHW 转成 HWC（matplotlib 显示图片的格式），再做最小-最大归一化，把数值缩放到 [0,1] 才能当彩色图显示；
- 用 4x8 的子图网格把 32 个滤波器全部画出来，每个标上编号 F0~F31；
- 注意：这里的模型未经训练，所以看到的是随机噪点；真正训练之后，这些滤波器会自动变成边缘、色块等模式的检测器。

````python
# Visualize first layer filters
filters = alexnet.conv1_filters  # Shape: (32, 3, 3, 3)

fig, axes = plt.subplots(4, 8, figsize=(16, 8))
axes = axes.flatten()

for i in range(min(32, len(axes))):
    # Normalize filter for visualization
    filt = filters[i].transpose(1, 2, 0)  # CHW -> HWC
    # min-max归一化到[0,1]便于显示;分母加1e-8防止滤波器为常数时除零
    filt = (filt - filt.min()) / (filt.max() - filt.min() + 1e-8)
    
    axes[i].imshow(filt)
    axes[i].axis('off')
    axes[i].set_title(f'F{i}', fontsize=8)

plt.suptitle('Conv1 Filters (32 filters, 3x3, RGB)', fontsize=14)
plt.tight_layout()
plt.show()

print("These filters learn to detect edges, colors, and simple patterns")
````

## Feature Map Visualization(特征图可视化)

#### 💻 代码解读

**做什么:** 把一张圆形图片送进第一层卷积，看看 32 个滤波器各自"看到"了什么——即可视化特征图（feature map）。

**怎么做:**
- 选取 `X_train[4]`（一张圆形图案）作为输入 `test_image`；
- 手动执行第一层的计算：`conv2d` 卷积 → 加偏置 `conv1_bias` → `relu` 激活，得到 32 张 32x32 的特征图 `conv1_output`；
- 用 6x6 的子图网格展示：第一格放原图，后面依次画出 32 张特征图（用 viridis 配色表示激活强弱）；
- 每张特征图就像一个滤波器的"视角"：有的对圆的边缘敏感、有的对整体亮度敏感——不同滤波器各司其职，这正是 CNN 分层提取特征的直观体现。

````python
# Process an image and visualize feature maps
test_image = X_train[4]  # Circle

# Forward through first conv layer
conv1_output = conv2d(test_image, alexnet.conv1_filters, stride=1, padding=1)
# 偏置广播:(32,) -> (32,1,1),加到每个特征图的所有位置
conv1_output += alexnet.conv1_bias[:, np.newaxis, np.newaxis]
conv1_output = relu(conv1_output)

# Visualize
fig = plt.figure(figsize=(16, 10))

# Original image
ax = plt.subplot(6, 6, 1)
ax.imshow(test_image.transpose(1, 2, 0))
ax.set_title('Input Image', fontsize=10)
ax.axis('off')

# Feature maps
for i in range(min(32, 35)):
    ax = plt.subplot(6, 6, i+2)
    # 每张特征图=一个卷积核在整幅图上的响应强度,亮处表示该模式被检测到
    ax.imshow(conv1_output[i], cmap='viridis')
    ax.set_title(f'Map {i}', fontsize=8)
    ax.axis('off')

plt.suptitle('Feature Maps after Conv1 + ReLU', fontsize=14)
plt.tight_layout()
plt.show()

print("Different feature maps respond to different patterns in the image")
````

## Test Classification(测试分类)

#### 💻 代码解读

**做什么:** 用（未训练的）AlexNet 对 10 张测试图片做分类预测，把类别分数换算成概率并展示预测结果。

**怎么做:**
- 定义 `softmax(x)`：先减去最大值（防止指数运算溢出），再取指数并归一化，把 10 个原始分数变成加起来等于 1 的概率——就像把"得分"换算成"各选项的把握程度"；
- 从测试集 `X_test` 每隔 5 张取一张，共 10 张；每张调用 `alexnet.forward(img, use_dropout=False)` 得到 logits，过 `softmax` 得概率 `probs`，用 `np.argmax` 选概率最大的类别作为预测 `pred_label`；
- 画出 2x5 网格，每格显示图片、真实标签（True）、预测标签（Pred）和置信度（Conf）；
- 最后打印提醒：模型未经训练、权重是随机的，所以预测基本靠"瞎猜"（约 10% 准确率）——完整训练需要反向传播和梯度下降，本笔记本为清晰起见做了简化。

````python
def softmax(x):
    # softmax数值稳定技巧:先减去最大值再取指数,防止exp溢出且结果不变
    exp_x = np.exp(x - np.max(x))
    return exp_x / exp_x.sum()

# Test on a few images
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
axes = axes.flatten()

for i in range(10):
    idx = i * 5  # Sample every 5th image
    img = X_test[idx]
    true_label = y_test[idx]
    
    # Forward pass
    # 推理时关闭dropout:训练才需要随机丢弃,预测要用完整网络
    logits = alexnet.forward(img, use_dropout=False)
    probs = softmax(logits)
    # argmax取概率最大的类别作为预测结果
    pred_label = np.argmax(probs)
    
    # Display
    axes[i].imshow(img.transpose(1, 2, 0))
    axes[i].set_title(f'True: {class_names[true_label]}\nPred: {class_names[pred_label]}\nConf: {probs[pred_label]:.2f}',
                     fontsize=9)
    axes[i].axis('off')

plt.suptitle('AlexNet Predictions (Untrained)', fontsize=14)
plt.tight_layout()
plt.show()

print("Note: Model is untrained, so predictions are random!")
print("Training would require gradient descent, which we've simplified for clarity.")
````

## Key Takeaways(关键要点)

### AlexNet Innovations (2012):(AlexNet 的创新（2012）：)

1. **ReLU Activation**: Much faster than sigmoid/tanh
   - No saturation for positive values
   - Faster training (6x compared to tanh)

2. **Dropout**: Powerful regularization
   - Prevents overfitting
   - Used in FC layers (0.5 rate)

3. **Data Augmentation**: 
   - Random crops and flips
   - Color jittering
   - Artificially increases dataset size

4. **GPU Training**: 
   - Used 2 GTX 580 GPUs
   - Enabled training of deep networks

5. **Local Response Normalization (LRN)**:
   - Lateral inhibition between feature maps
   - Less common now (Batch Norm replaced it)

1. **ReLU 激活函数(ReLU Activation)**：比 sigmoid/tanh 快得多
   - 正值不会饱和
   - 训练更快（相比 tanh 快 6 倍）

2. **Dropout**：强大的正则化(regularization)手段
   - 防止过拟合(overfitting)
   - 用于全连接层（丢弃率 0.5）

3. **数据增强(Data Augmentation)**：
   - 随机裁剪与翻转
   - 颜色抖动(color jittering)
   - 人为扩大数据集规模

4. **GPU 训练**：
   - 使用了 2 块 GTX 580 GPU
   - 使深度网络的训练成为可能

5. **局部响应归一化(Local Response Normalization, LRN)**：
   - 特征图之间的侧向抑制
   - 如今已不常用（被批归一化 Batch Norm 取代）

```
Input (227x227x3)

  ↓

  ↓

FC7 (4096) + ReLU + Dropout
  ↓
FC8 (1000 classes) + Softmax
```
